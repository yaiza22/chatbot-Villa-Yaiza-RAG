# Imports

```python
import sys
from pathlib import Path
import pandas as pd

# Configurar ruta para importar la carpeta src
sys.path.append(str(Path.cwd().parent))

from src.config import PROCESSED_DIR
from src.engines.m0_rules import RuleEngine
from src.engines.m1_tfidf_rl import TfidfRlEngine

print("Librerías y módulos cargados correctamente.")
```

# Training

```python
# Cargar datos y entrenar M1
print("Cargando datos reales...")
labeled = pd.read_csv(PROCESSED_DIR / "labeled.csv")
train_df = labeled[labeled["split"] == "train"]

print(f"Datos de entrenamiento cargados: {len(train_df)} preguntas reales.\n")

print("Inicializando motores...")
m0 = RuleEngine()
m1 = TfidfRlEngine()

m0.fit(train_df)
print("Entrenando M1 (esto toma unos segundos)...")
m1.fit(train_df)
print("¡Motores listos para la demostración!")
```

# Demo function

```python
def probar_bot(pregunta):
    print(f"\n🙋‍♂️ Cliente: '{pregunta}'")
    print("-" * 60)
    
    res_m0 = m0.respond(pregunta)
    print(f"🤖 M0 (Reglas):")
    print(f"   - Intención: {res_m0.intent}")
    print(f"   - ¿Escala a humano?: {res_m0.escalate}")
    
    res_m1 = m1.respond(pregunta)
    print(f"🧠 M1 (Machine Learning):")
    print(f"   - Intención: {res_m1.intent}")
    print(f"   - ¿Escala a humano?: {res_m1.escalate}")
    print(f"   - Confianza: {res_m1.confidence:.2f}")
    print("=" * 60)
```

# Live test

```python
# Ejecutar la demostración
probar_bot("buenas tardes cuanto vale la finca para el 24 de diciembre?")
probar_bot("cuantas personas caben en la finca?")
probar_bot("se permiten mascotas?")
probar_bot("me recomiendas un hotel en cartagena?")
```
